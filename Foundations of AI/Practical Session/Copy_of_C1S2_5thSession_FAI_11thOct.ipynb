```python
# Facts: What we know about animals
facts = {
    'dog': {'mammal': True, 'gives_live_birth': True},
    'cat': {'mammal': True, 'gives_live_birth': True},
    'fish': {'mammal': False, 'gives_live_birth': False},
    'bird': {'mammal': False, 'gives_live_birth': False},
}

# Rule: If an animal is a mammal and gives live birth, it's warm-blooded.
def is_warm_blooded(animal):
    if facts.get(animal):
        if facts[animal].get('mammal') and facts[animal].get('gives_live_birth'):
            return True
    return False

# Reasoning: Let's check if certain animals are warm-blooded
animals_to_check = ['dog', 'cat', 'fish', 'bird']

for animal in animals_to_check:
    if is_warm_blooded(animal):
        print(f"The {animal} is warm-blooded.")
    else:
        print(f"The {animal} is not warm-blooded.")
```

```python
import gradio as gr

# Initialize an empty knowledge base (facts)
facts = {}

# Function to handle adding facts
def add_fact(animal, characteristic, value):
    facts[animal] = facts.get(animal, {})
    facts[animal][characteristic] = value
    return f"Fact added: {animal} is {characteristic} = {value}"

# Rule: If an animal is a mammal and gives live birth, it's warm-blooded.
def is_warm_blooded(animal):
    if facts.get(animal):
        if facts[animal].get('mammal') and facts[animal].get('gives_live_birth'):
            return True
    return False

# Function to handle reasoning based on the facts
def reason_about_animals():
    results = []
    if not facts:
        return "No facts available. Please add some facts first."

    for animal in facts:
        if is_warm_blooded(animal):
            results.append(f"The {animal} is warm-blooded.")
        else:
            results.append(f"The {animal} is not warm-blooded.")

    return "\n".join(results)

# Create the Gradio interface
def create_gradio_interface():
    with gr.Blocks() as demo:
        gr.Markdown("# Animal Knowledge Representation and Reasoning")

        with gr.Tab("Add Fact"):
            animal_input = gr.Textbox(label="Animal Name", placeholder="e.g., dog, cat", interactive=True)
            characteristic_input = gr.Dropdown(
                label="Characteristic", choices=["mammal", "gives_live_birth"], interactive=True
            )
            value_input = gr.Dropdown(label="Value", choices=["True", "False"], interactive=True)
            add_fact_button = gr.Button("Add Fact")
            fact_output = gr.Textbox(label="Fact Status", interactive=False)

            add_fact_button.click(
                add_fact,
                inputs=[animal_input, characteristic_input, value_input],
                outputs=fact_output
            )

        with gr.Tab("Reasoning"):
            reason_button = gr.Button("Reason About Animals")
            reasoning_output = gr.Textbox(label="Reasoning Results", interactive=False)

            reason_button.click(reason_about_animals, inputs=[], outputs=reasoning_output)

    demo.launch()

# Run the Gradio app
create_gradio_interface()
```
