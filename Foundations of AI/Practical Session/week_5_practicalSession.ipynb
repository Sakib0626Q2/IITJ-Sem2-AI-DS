```python
# rule: animal can walk
# fact: dog is animal

#dictonary, key walue pair

rule = {
    "animal" : "can walk",
    "bird" : "can fly",
    "fish" : "can swim"
}

facts = {
    "dog" : "animal",
    "sparrow" : "bird",
    "dolphin" : "fish"
}

print(rule[facts["dog"]])
print(rule[facts["sparrow"]])
print(rule[facts["dolphin"]])
```
