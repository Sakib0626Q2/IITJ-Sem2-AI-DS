```python
# Under shooting

import matplotlib.pyplot as plt
import numpy as np

def f(x):
  return (x-3)**2 + 4

def df(x):
  return 2 * (x-3)

xStart = -4
lr = 0.01
epochs = 50

x = xStart
xHistory = [x]

xPlot = np.linspace(-5, 10, 100)
y = f(xPlot)

for epoch in range(epochs):
  gradient = df(x)
  x = x - lr * gradient
  xHistory.append(x)

plt.plot(xPlot, y, label="f(x) = (x-3)^2 + 4")
plt.scatter(xHistory, [f(i) for i in xHistory], color="red", label="Descent Steps")
plt.scatter(3, 4, color="green", label = "True Minimum")

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Plot of f(x)")
plt.grid(True)
plt.legend()
plt.show()

print({"Final x: ": x,
       "Iterations": epochs})
```

# Part B – Activity: Visualizing Gradient Descent

```python
LR = [0.01, 0.1, 0.9]
record = []

for lr in LR:
  xStart = -4
  epochs = 50
  plt.figure(figsize=(7,5))

  x = xStart
  xHistory = [x]

  xPlot = np.linspace(-5, 10, 100)
  y = f(xPlot)

  for epoch in range(epochs):
    gradient = df(x)
    x = x - lr * gradient
    xHistory.append(x)

  plt.plot(xPlot, y, label="f(x) = (x-3)^2 + 4")
  plt.scatter(xHistory, [f(i) for i in xHistory], color="red", label="Descent Steps")
  plt.plot(xHistory, [f(i) for i in xHistory], color="red")

  plt.scatter(3, 4, color="green", label = "True Minimum")

  plt.xlabel("x")
  plt.ylabel("f(x)")
  plt.title(f"Gradient Descent (lr = {lr})")
  plt.grid(True)
  plt.legend()
  plt.show()

  values = {"learning_rate": lr,
            "Final x: ": x,
            "Iterations": epochs}
  record.append(values)
print(record)
```
