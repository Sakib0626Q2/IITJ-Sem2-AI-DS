```python
# -------------------------------------------------------
# Comparison: SGD vs Mini-Batch Gradient Descent
# -------------------------------------------------------

import numpy as np
import matplotlib.pyplot as plt

# For reproducibility of random numbers
np.random.seed(42)

# -------------------------------------------------------
# STEP 1: Create synthetic data (same as Lab 1)
# -------------------------------------------------------
# y = 2 + 3x + noise

X = 2 * np.random.rand(1000, 1)         # 1000 samples of single feature between 0 and 2
y = 2 + 3 * X + np.random.randn(1000, 1)  # Linear relation with Gaussian noise

# Add bias term (x0 = 1) to each input sample → X_b = [1, x]
X_b = np.c_[np.ones((1000, 1)), X]       # shape: (1000, 2)

# -------------------------------------------------------
# STEP 2: Initialize variables
# -------------------------------------------------------

# Random initial weights for both methods: [bias, slope]
weights_sgd = np.random.randn(2, 1)
weights_mbgd = np.random.randn(2, 1)

# Learning rate and other parameters
lr = 0.05            # step size to 0.1
epochs = 30          # how many passes over full dataset epochs to 100
m = len(X_b)         # number of samples
batch_size = 32      # mini-batch size batch sizes to 64 and caculate again

# Lists to track loss over epochs
loss_sgd = []
loss_mbgd = []

# -------------------------------------------------------
# STEP 3: Train using Stochastic Gradient Descent (SGD)
# -------------------------------------------------------
# In SGD, we update the weights after *each* sample.
# That means: gradient is computed using only 1 data point.

print("==== Training with Stochastic Gradient Descent (SGD) ====\n")

for epoch in range(epochs):
    # Shuffle the data each epoch for better convergence
    indices = np.random.permutation(m)
    X_b_shuffled = X_b[indices]
    y_shuffled = y[indices]

    for i in range(m):
        xi = X_b_shuffled[i:i+1]   # single training example (row)
        yi = y_shuffled[i:i+1]     # corresponding label

        # Prediction for this single example
        pred = xi.dot(weights_sgd)

        # Error for this single prediction
        error = pred - yi

        # Compute gradient of MSE loss: grad = 2 * xᵀ * error
        grad = 2 * xi.T.dot(error)

        # Update rule: w := w - η * grad
        weights_sgd -= lr * grad

        # Print the first few gradients for observation
        if i < 5:
            print(f"Epoch {epoch+1}, Sample {i+1} Gradient: {grad.flatten()}")

    # Compute MSE loss for the entire dataset after epoch
    preds_all = X_b.dot(weights_sgd)
    loss = ((preds_all - y) ** 2).mean()
    loss_sgd.append(loss)

    print(f"Epoch {epoch+1} - SGD Loss: {loss:.4f}\n")

# -------------------------------------------------------
# STEP 4: Train using Mini-Batch Gradient Descent (MBGD)
# -------------------------------------------------------
# In MBGD, we update weights after each *mini-batch* (subset of data).
# Batch size is usually 16, 32, or 64.

print("\n==== Training with Mini-Batch Gradient Descent (MBGD) ====\n")

for epoch in range(epochs):
    # Shuffle the dataset at the start of each epoch
    indices = np.random.permutation(m)
    X_b_shuffled = X_b[indices]
    y_shuffled = y[indices]

    # Loop through mini-batches
    for start in range(0, m, batch_size):
        end = start + batch_size
        xb = X_b_shuffled[start:end]
        yb = y_shuffled[start:end]

        # Predictions for this mini-batch
        preds = xb.dot(weights_mbgd)

        # Compute errors
        error = preds - yb

        # Gradient averaged over the mini-batch
        grad = (2 / xb.shape[0]) * xb.T.dot(error)

        # Print first few batches and their gradients
        print(f"Epoch {epoch+1}, Mini-batch indices: {start} to {end-1}")
        print(f"Mini-batch Gradient (first 2 elements): {grad.flatten()[:2]}")

        # Weight update
        weights_mbgd -= lr * grad

    # Compute loss on entire dataset after each epoch
    preds_all = X_b.dot(weights_mbgd)
    loss = ((preds_all - y) ** 2).mean()
    loss_mbgd.append(loss)

    print(f"Epoch {epoch+1} - Mini-Batch GD Loss: {loss:.4f}\n")

# -------------------------------------------------------
# STEP 5: Plot comparison of convergence
# -------------------------------------------------------

plt.plot(range(1, epochs + 1), loss_sgd, marker='o', label='SGD')
plt.plot(range(1, epochs + 1), loss_mbgd, marker='s', label='Mini-Batch GD')
plt.xlabel('Epoch')
plt.ylabel('Loss (MSE)')
plt.title('Convergence: SGD vs Mini-Batch Gradient Descent')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Lab 3 Momentum Gradient Descent

weights = np.random.randn(2, 1)  #intial random weights
velocity = np.zeros_like(weights) # initial velocity (same shape as weights)

lr = 0.1 # learning rate
beta = 0.9 # smothness (momentum coefficient)
epochs = 50  #iterations
loss_history = []   #to track loss per epoch


for epoch in range(epochs):

  preds = X_b.dot(weights) #forward pass:predictions for all samples
  error = preds - y  #computing error
  grad = (2/m) * X_b.T.dot(error)  # Gradient wrt MSE
  velocity = beta * velocity + lr * grad # updated velocity
  weights -= lr * velocity #updated weights
  loss = np.mean(error ** 2) #computing MSE Loss
  loss_history.append(loss)
  print(f"Epoch {epoch + 1} Loss: {loss:.4f}")

  if epoch < 5:  #printing intermediate results for fist few epochs
    print(f"Epoch {epoch + 1} Gradient: {grad.flatten()} Velocity: {velocity.flatten()} Loss: {loss:.4f}")

plt.plot(range(1, epochs + 1), loss_history, marker='o')
plt.xlabel('Epochs')
plt.ylabel('Loss (MSE)')
plt.title('Momentum Gradient Descent Loss Curve')
plt.show()


```
