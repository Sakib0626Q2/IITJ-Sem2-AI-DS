```python
# Prompt Chaining

# Step 1: Extract keywords
prompt1 = "Extract keywords from: AI is transforming healthcare"

res1 = client.chat.completions.create(
    model="llama-3.1-8b-instant",
    messages=[{"role": "user", "content": prompt1}]
)

keywords = res1.choices[0].message.content

# Step 2: Summarize using keywords
prompt2 = f"Write a short summary using: {keywords}"

res2 = client.chat.completions.create(
    model="llama-3.1-8b-instant",
    messages=[{"role": "user", "content": prompt2}]
)

print(res2.choices[0].message.content)
```
