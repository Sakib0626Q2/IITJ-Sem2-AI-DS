```python
import pandas as pd

data = {
    "Architecture": [
        "3000→1000→500→10",
        "3000→500→10",
        "3000→10",
        "2000→10",
        "1500→10",
        "784→128→10",
        "784→128→64→10"
    ],
    "Accuracy (%)": [
        97.86,
        98.22,
        97.85,
        98.16,
        98.01,
        98.08,
        97.84
    ],
    "Test Loss": [
        0.1251,
        0.1071,
        0.1071,
        0.1032,
        0.1047,
        0.0883,
        0.0907
    ],
    "Time per Epoch": [
        66,
        40,
        22,
        11,
        6,
        2,
        2
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
fig, ax1 = plt.subplots(figsize=(12, 6))

# Accuracy plot
ax1.plot(df["Architecture"], df["Accuracy (%)"], marker='o', label='Accuracy')
ax1.set_xlabel("Architecture")
ax1.set_ylabel("Accuracy (%)")
ax1.tick_params(axis='x', rotation=20)

# Loss plot on same axis
ax1.plot(df["Architecture"], df["Test Loss"], marker='s', label='Loss')

# Second axis for time
ax2 = ax1.twinx()
ax2.plot(df["Architecture"], df["Time per Epoch"], marker='^', linestyle='--', label='Time (s)')
ax2.set_ylabel("Time per Epoch (s)")

# Combined legend
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()

ax1.legend(lines1 + lines2, labels1 + labels2, loc='best')

plt.title("Architecture vs Accuracy, Loss, and Training Time")
plt.tight_layout()
plt.show()
```

```python
from sklearn.preprocessing import MinMaxScaler
normalized_df = df.copy()

for col in ["Accuracy (%)", "Test Loss", "Time per Epoch"]:
    normalized_df[col] = (
        (df[col] - df[col].min()) /
        (df[col].max() - df[col].min())
    )

print(normalized_df)
# cols = ["Accuracy (%)", "Test Loss", "Time per Epoch"]

# # Apply MinMaxScaler
# scaler = MinMaxScaler()

# normalized_df[cols] = scaler.fit_transform(normalized_df[cols])

# print(normalized_df)
```

```python
plt.figure(figsize=(12,6))

plt.plot(
    normalized_df["Architecture"],
    normalized_df["Accuracy (%)"],
    marker='o',
    linewidth=2,
    label="Accuracy ↑"
)

plt.plot(
    normalized_df["Architecture"],
    normalized_df["Test Loss"],
    marker='s',
    linewidth=2,
    label="Loss ↓"
)

plt.plot(
    normalized_df["Architecture"],
    normalized_df["Time per Epoch"],
    marker='^',
    linewidth=2,
    linestyle='--',
    label="Time ↓"
)

plt.xlabel("Architecture")
plt.ylabel("Normalized Value")
plt.title("Normalized Architecture Comparison")
plt.xticks(rotation=20)
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```
