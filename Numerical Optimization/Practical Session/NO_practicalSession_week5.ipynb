# Gradient Desent Family

#### Batch GD : 
 - Updates parameters using the entire dataset
 - Full Crowd
 - speed - lowest; accuracy - Highest
 - eg: Avg height of a class

#### Stochastic GD :
 - Updates parameters using one sample at a time
 - single person
 - speed - Highest; accuracy - lowest
 - eg: fine for quick/lightning fast answers, no thinking; It may be wrong but I would have a head start

#### Mini-batch GD : 
 - Updates parameters using a small batch of samples
 - Mix of the two above
 - eg: balance between speed and accuracy

```python
# Batch Gradient Descent for Linear Regression

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(40)  # For reproducibility

# Generate synthetic data y = 3x + 2 +noise
X = 2 * np.random.rand(1000, 1)
y = 3 * X + 2 + np.random.randn(1000, 1) * 0.5

#add bias term to (column of ones)
X_b = np.c_[np.ones((1000, 1)), X]

#initialize weights randomly
weights = np.random.randn(2, 1)

#learning rate
lr = 0.1
epoach = 50
m = len(X_b)
loss_history = []

for epoch in range(epoach):
    # Compute predictions
    pred = X_b.dot(weights)

    #Compute error
    error = pred - y

    # Compute gradients
    gradients = (2/m) * X_b.T.dot(error)
    
    # Compute loss (Mean Squared Error)
    loss = np.mean((pred - y) ** 2)
    loss_history.append(loss)

    # Update weights
    weights -= lr * gradients
    print(f"Epoch {epoch+1}/{epoach}, Loss: {loss:.4f}")

# Plot loss curve
plt.plot(range(1, epoach + 1), loss_history, marker='o')
plt.xlabel('Epoch')
plt.ylabel('Loss (MSE)')
plt.title('Loss Curve for Batch Gradient Descent')
plt.show()
```
