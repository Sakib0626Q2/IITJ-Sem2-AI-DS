```python
# YOUR CODE HERE
splitPoint = int(0.8 * len(df))

X_train = scaled_features[:splitPoint]
X_test = scaled_features[splitPoint:]

y_train = df['Class_Label'][:splitPoint]
y_test = df['Class_Label'][splitPoint:]

# df.info()
```

### Q3: Euclidean Distance

Write a function `calculate_euclidean_distance(a, b)`
which returns the Euclidean distance between two numpy vectors


```python
# YOUR CODE HERE
def calculate_euclidean_distance(a, b):
    return np.sqrt(np.sum((a-b)**2))
```

### Q4: Single Prediction Using KNN

Write a function `knn_predict_single(test_point, X_train, y_train, K=5)` that:
1. Computes distances between test_point and all training samples
2. Finds the K nearest neighbors
3. Returns the predicted class label based on majority vote


```python
# YOUR CODE HERE
def knn_predict_single(test_point, X_train, y_train, K=5):
    distances = []
    k_nearest_labels = []
    for i in range(len(X_train)):
        dist = calculate_euclidean_distance(test_point, X_train[i])
        distances.append((dist, y_train.iloc[i]))
    
    distances.sort(key=lambda x: x[0])

    for i in range(K):
        k_nearest_labels.append(distances[i][1])
    
    most_common = Counter(k_nearest_labels).most_common(1)
    return most_common[0][0]
```

### Q5: Full Test Prediction

Use `knn_predict_single` to generate predictions for entire X_test.
Store predictions list in variable: predictions


```python
# YOUR CODE HERE
predictions = []
for test_row in X_test:
    result = knn_predict_single(test_row, X_train, y_train, K=5)
    predictions.append(result)

print(predictions)
```

### Q6: Accuracy Score

Compute accuracy and store in variable `accuracy_score`


```python
# YOUR CODE HERE
correct_count = 0
total_count = len(y_test)

y_test_list = y_test.tolist()

for i in range(total_count):
    if predictions[i] == y_test_list[i]:
        correct_count += 1

accuracy_score = correct_count / total_count
print("The accuracy of the model is:", accuracy_score)
```
