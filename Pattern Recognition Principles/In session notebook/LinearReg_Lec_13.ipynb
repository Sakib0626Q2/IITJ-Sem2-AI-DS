```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Step 1: Toy Dataset
# -----------------------------
X = np.array([1,2,3,4,5], dtype=float)
y = np.array([4,8,10,12,15], dtype=float)

# -----------------------------
# Step 2: OLS (Manual Formula)
# -----------------------------
x_mean = np.mean(X)
y_mean = np.mean(y)

# slope m and intercept c
m = np.sum((X - x_mean) * (y - y_mean)) / np.sum((X - x_mean)**2)
c = y_mean - m * x_mean

print("OLS Solution:")
print(f"  m = {m:.4f}, c = {c:.4f}")

# -----------------------------
# Step 3: Gradient Descent
# -----------------------------
learning_rate = 0.01
n_iterations = 1000

m_gd = np.random.randn()
c_gd = np.random.randn()

n = len(X)

for _ in range(n_iterations):
    y_pred = m_gd * X + c_gd

    dm = (-2/n) * np.sum(X * (y - y_pred))
    dc = (-2/n) * np.sum(y - y_pred)

    m_gd -= learning_rate * dm
    c_gd -= learning_rate * dc

print("\nGradient Descent Solution:")
print(f"  m = {m_gd:.4f}, c = {c_gd:.4f}")

# -----------------------------
# Step 4: Side-by-side plots
# -----------------------------
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

# OLS Plot
ax1.scatter(X, y, label="Data Points", color="black")
ax1.plot(X, m*X + c, label="OLS Fit", linewidth=3)
ax1.set_title("OLS Regression Line")
ax1.set_xlabel("X")
ax1.set_ylabel("Y")
ax1.legend()

# GD Plot
ax2.scatter(X, y, label="Data Points", color="black")
ax2.plot(X, m_gd*X + c_gd, label="GD Fit", linewidth=3, linestyle='--')
ax2.set_title("Gradient Descent Regression Line")
ax2.set_xlabel("X")
ax2.set_ylabel("Y")
ax2.legend()

plt.tight_layout()
plt.show()
```
