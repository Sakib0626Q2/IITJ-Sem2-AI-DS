```python
fig, axs = plt.subplots(1, 2, figsize=(10,8))
axs[0].bar(df['Attrition'].value_counts().index, df['Attrition'].value_counts().values)
axs[0].set_xlabel('Attrition')
axs[0].set_ylabel('Count')
# axs[0].set_title('Employee Attrition Distribution')

axs[1].pie(df['Attrition'].value_counts().values, labels=df['Attrition'].value_counts().index, autopct='%1.1f%%')
# axs[1].set_title('Employee Attrition Distribution (Pie Chart)')

plt.suptitle('Employee Attrition Distribution')
plt.axis('equal')
plt.show()

# plt.bar(df['Attrition'].value_counts().index, df['Attrition'].value_counts().values)
# plt.xlabel('Attrition')
# plt.ylabel('Count')
# plt.title('Employee Attrition Distribution')
```

```python
plt.scatter(df['Age'], df['MonthlyIncome'], color="#6edefaa0")
plt.xlabel('Age')
plt.ylabel('Monthly Income')
plt.title('Age vs Monthly Income')
plt.show()
```

```python
df['Attrition'] = df['Attrition'].map({'Yes':1, 'No':0})
corr = df.corr(numeric_only=True)['Attrition'].sort_values(ascending=False)
print(corr)
```

```python
plt.figure(figsize=(10,6))
plt.bar(corr.index, corr.values)
plt.xticks(rotation=90)
plt.xlabel('Features')
plt.ylabel('Correlation with Attrition')
plt.title('Correlation of Features with Attrition')
plt.show()
```

```python
corr_top = corr.abs().sort_values()
corr_top = corr_top[corr_top < 0.1]

corr_top.plot(kind='barh', figsize=(8,6))
plt.title("Features correlated with Attrition")
plt.show()
```

```python
corr_top = corr.abs().sort_values()
corr_top = corr_top[corr_top > 0.1]

corr_top.plot(kind='barh', figsize=(8,6))
plt.title("Features correlated with Attrition")
plt.show()
```
