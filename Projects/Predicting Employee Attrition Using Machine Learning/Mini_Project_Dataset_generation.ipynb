```python
import pandas as pd
import numpy as np

# Set random seed so the dataset is reproducible
np.random.seed(42)

n_rows = 500

# 1. Generate Independent Variables based on your constraints
employee_id = np.arange(1001, 1001 + n_rows)
age = np.random.randint(22, 59, n_rows)
gender = np.random.choice(['Male', 'Female'], n_rows, p=[0.55, 0.45])
department = np.random.choice(['IT', 'Sales', 'HR', 'Finance', 'R&D'], n_rows)

# Map realistic roles based on the assigned department
def assign_role(dept):
    roles = {
        'IT': ['Software Engineer', 'Data Scientist', 'System Admin'],
        'Sales': ['Sales Executive', 'Account Manager'],
        'HR': ['HR Generalist', 'Recruiter'],
        'Finance': ['Financial Analyst', 'Accountant'],
        'R&D': ['Research Scientist', 'Lab Technician']
    }
    return np.random.choice(roles[dept])

job_role = [assign_role(d) for d in department]

# Ensure YearsAtCompany makes logical sense with Age (assuming start at 22)
max_years_possible = age - 22
years_at_company = [np.random.randint(0, min(21, max_y + 1)) for max_y in max_years_possible]

# Ensure YearsSinceLastPromotion doesn't exceed YearsAtCompany
years_since_promotion = [np.random.randint(0, min(11, yac + 1)) for yac in years_at_company]

monthly_income = np.random.randint(15000, 150001, n_rows)
job_satisfaction = np.random.randint(1, 5, n_rows)
work_life_balance = np.random.randint(1, 5, n_rows)
education_level = np.random.randint(1, 6, n_rows)
training_hours = np.random.randint(0, 81, n_rows)
overtime = np.random.choice(['Yes', 'No'], n_rows, p=[0.3, 0.7])
performance_rating = np.random.randint(1, 6, n_rows)

# 2. Generate Target Variable (Attrition) with Built-in Correlations
attrition_prob = np.full(n_rows, 0.10) # 10% base chance of leaving

for i in range(n_rows):
    # Factors that push employees out
    if overtime[i] == 'Yes': attrition_prob[i] += 0.25
    if job_satisfaction[i] <= 2: attrition_prob[i] += 0.20
    if work_life_balance[i] <= 2: attrition_prob[i] += 0.15
    if years_since_promotion[i] > 4: attrition_prob[i] += 0.10

    # Factors that retain employees
    if job_satisfaction[i] >= 4: attrition_prob[i] -= 0.15
    if monthly_income[i] > 100000: attrition_prob[i] -= 0.10

# Keep probabilities within valid bounds
attrition_prob = np.clip(attrition_prob, 0.05, 0.90)

# Assign Yes/No based on the calculated probability
attrition = [np.random.choice(['Yes', 'No'], p=[p, 1-p]) for p in attrition_prob]

# 3. Assemble and Export
df = pd.DataFrame({
    'EmployeeID': employee_id,
    'Age': age,
    'Gender': gender,
    'Department': department,
    'JobRole': job_role,
    'MonthlyIncome': monthly_income,
    'YearsAtCompany': years_at_company,
    'YearsSinceLastPromotion': years_since_promotion,
    'JobSatisfaction': job_satisfaction,
    'WorkLifeBalance': work_life_balance,
    'EducationLevel': education_level,
    'TrainingHoursLastYear': training_hours,
    'OverTime': overtime,
    'PerformanceRating': performance_rating,
    'Attrition': attrition
})

# Save to CSV
df.to_csv('TechNova_Employee_Attrition.csv', index=False)
print("Dataset generated successfully!")
```
