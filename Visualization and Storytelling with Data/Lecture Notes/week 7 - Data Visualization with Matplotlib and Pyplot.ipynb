```python
# Interactive plotting with Plotly and Matplotlib

import matplotlib.pyplot as plt
import plotly.express as px
import pandas as pd
import numpy as np

#create data
x = np.linspace(0, 3*np.pi, 100)
y = np.sin(x)

df = pd.DataFrame({'x': x, 'y': y})

# interactive lineplot
fig = px.line(df, x = 'x', y = 'y', title = 'Interactive Sin Plot')
fig.show()
```

```python
x = np.linspace(0, 3*np.pi, 100)
df = pd.DataFrame({
    'x': x,
    'sin(x)': np.sin(x),
    'cos(x)': np.cos(x)
    })

fig = px.line(df, x = 'x', y = ['sin(x)', 'cos(x)'], title = 'Interactive Sin and Cos Plot')
fig.show()
```

```python
x = np.linspace(0, 3*np.pi, 100)
df = pd.DataFrame({
    'x': x,
    'sin(x)': np.sin(x),
    'cos(x)': np.cos(x)
    })

fig = px.scatter(df, x = 'x', y = ['sin(x)', 'cos(x)'], title = 'Interactive Sin and Cos Plot')
fig.show()
```
