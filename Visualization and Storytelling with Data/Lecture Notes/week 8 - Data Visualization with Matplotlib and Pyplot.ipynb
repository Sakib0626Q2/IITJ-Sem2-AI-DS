```python
from plotly.subplots import make_subplots
import plotly.graph_objects as go
import numpy as np
import pandas as pd
```

```python
#create mixed subplot
fig = make_subplots(rows = 1, cols = 2, subplot_titles = ("line Plot", "Bar Chart"))
fig.add_trace(go.Scatter(x = [1, 2, 3, 4], y = [10, 15, 20, 25], mode = "lines", name = "Line"), row = 1, col = 1)
fig.add_trace(go.Bar(x = ['A', 'B', 'C', 'D'], y = [10, 15, 20, 25], name = "Bar"), row = 1, col = 2)
fig.update_layout(title_text = "Mixed Subplot Example", showlegend = True)
fig.show()
```

```python
import plotly.express as px

df = px.data.iris()
print(df.info())
```

```python
fig = px.scatter(df, x = "sepal_width", y = "sepal_length", color = "species", size = 'petal_length', hover_data=['petal_width'], title = "Iris Dataset Scatter Plot")
fig.show()
```

```python
data = {'fruit': ['Apple', 'Banana', 'Orange', 'Grape', 'Pinapple'],
        'Quantity': [120, 80, 150, 90, 60]}
df = pd.DataFrame(data)
fig = px.bar(df, x ='fruit', y = 'Quantity',text = 'Quantity',color ='fruit' ,title = "Fruit Quantity Bar Chart")
fig.update_traces(textposition = 'outside')
fig.update_layout(showlegend = True)
fig.show()
```
