# Interactive Heat Map

```python
import plotly.express as px
import pandas as pd
import numpy as np
from plotly.subplots import make_subplots
import plotly.graph_objects as go


data = np.random.rand(10, 10)
df = pd.DataFrame(data, columns=[f'Column {i}' for i in range(1, 11)])

print(df)
```

```python
fig = px.imshow(df, color_continuous_scale='Inferno', title='Interactive Heatmap')
fig.show()
```

Dashboard

```python
x = np.linspace(0, 4*np.pi, 100)

y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.exp(-x)

fig = make_subplots(rows = 2, cols = 2, subplot_titles = ('Sine Wave', 'Cosine  Wave', 'Exp Decay', 'sin(x) + cos(x)'))

fig.add_trace(go.Scatter(x = x, y = y1, mode = 'lines', name = 'sin(x)'), row = 1, col = 1)
fig.add_trace(go.Scatter(x = x, y = y2, mode = 'lines', name = 'cos(x)'), row = 1, col = 2)
fig.add_trace(go.Scatter(x = x, y = y3, mode = 'lines', name = 'exp(-x)'), row = 2, col = 1)
fig.add_trace(go.Scatter(x = x, y = y1+y2, mode = 'lines', name = 'sin(x) + cos(x)'), row = 2, col = 2)

fig.update_layout(height = 700, width = 900, title_text = 'Interactive Subplots Dashboard')
fig.show()
```
